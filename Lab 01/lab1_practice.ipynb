```python
import re
text="Hello!!! Welcome <br> to the NLP Lab 2026."
temp=re.sub(r'<[^>]+>','',text)
print(temp)
temp1=re.sub(r'[^A-Za-z\s]','',temp)
print(temp1)
temp2=re.sub(r'\s+',' ',temp1)
print(temp2)
```

```python
from nltk.tokenize import word_tokenize
tokens=word_tokenize(temp2.lower())
print(tokens)
```

```python
from nltk.corpus import stopwords
s=set(stopwords.words("english"))
filtered=[]
for w in tokens:
    if w not in s: 
        filtered.append(w)
print(filtered)
```

```python
from nltk.stem import PorterStemmer
s=PorterStemmer()
words=["run","running","ran","easily","fairly"]
for w in words:
    print(w,":",s.stem(w))
```

```python
from nltk.stem import WordNetLemmatizer
l=WordNetLemmatizer()
words=["run","running","ran","easily","fairly"]
for w in words:
    print(w,":",l.lemmatize(w,pos='v')) 
```
