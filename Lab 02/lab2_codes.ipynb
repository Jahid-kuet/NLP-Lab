# 3. Text Preprocessing

## Function

```python
def full_preprocess(text,
                    remove_stopwords=True,
                    use_lemmatization=True):

    text = text.lower()

    text = re.sub(r'<[^>]+>', '', text)

    text = re.sub(r'[^a-z\s]', '', text)

    tokens = word_tokenize(text)

    if remove_stopwords:
        stop_words = set(stopwords.words('english'))
        tokens = [word for word in tokens if word not in stop_words]

    if use_lemmatization:
        lemmatizer = WordNetLemmatizer()
        tokens = [lemmatizer.lemmatize(word) for word in tokens]

    return tokens
```

```python
    print(full_preprocess(sentence))
```

# 6. Build N-Gram Model

```python
def build_ngram_model(tokens, n=2):

    model = defaultdict(Counter)

    for i in range(len(tokens) - n + 1):

        history = tuple(tokens[i:i+n-1])

        next_word = tokens[i+n-1]

        model[history][next_word] += 1

    probability_model = defaultdict(dict)

    for history, counter in model.items():

        total = sum(counter.values())

        for word, frequency in counter.items():

            probability_model[history][word] = frequency / total

    return probability_model
```

# 10. Shannon's Next Word Predictor

## Function

```python
def shannons_predictor(history_phrase,
                       model,
                       n):

    tokens = full_preprocess(history_phrase,
                             remove_stopwords=False)

    history = tuple(tokens[-(n-1):])

    if history not in model:
        return []

    predictions = sorted(model[history].items(),
                         key=lambda x: x[1],
                         reverse=True)

    return predictions
```

# 11. Random Text Generation

## Function

```python
def generate_text(model,
                  seed_phrase,
                  max_words,
                  n):

    output_tokens = full_preprocess(seed_phrase,
                                    remove_stopwords=False)

    for i in range(max_words):

        history = tuple(output_tokens[-(n-1):])

        if history not in model:
            break

        words = list(model[history].keys())

        probabilities = list(model[history].values())

        next_word = np.random.choice(words,
                                     p=probabilities)

        output_tokens.append(next_word)

    return " ".join(output_tokens)
```

## Example (Bigram)

```python
print(generate_text(bigram_model,
                    "quick",
                    10,
                    2))
```

## Example (Trigram)

```python
print(generate_text(trigram_model,
                    "the quick",
                    10,
                    3))
```
